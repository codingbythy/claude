# Stargazer's Cabin: make the long video
Press the play button on the cell below. It builds the video in Google's cloud and saves it into your Google Drive, so nothing is stored on your phone or computer.


```
# Builds the Stargazer's Cabin video and saves it straight into your Google Drive.
# Press the play button on the left, allow Google Drive access, and wait about 5 minutes.

LOOPS = 40                       # 40 loops = about 2 hours (each loop is 3 minutes). 60 = 3 hours.
DRIVE_FOLDER = "Stargazers Cabin"  # folder in My Drive where the video is saved

import os, shutil, subprocess
from google.colab import drive
drive.mount("/content/drive")

BASE = "https://raw.githubusercontent.com/codingbythy/claude/refs/heads/claude/focused-curie-orkd4f/cozy-cabin/video-kit/"
os.makedirs("/content/kit", exist_ok=True)
os.chdir("/content/kit")
for name in ["intro.mp4", "loop.mp4", "outro.mp4", "ambience.flac", "music.flac", "build.sh"]:
    print("Downloading", name)
    subprocess.run(["wget", "-q", "-O", name, BASE + name], check=True)

hours = round((170 + LOOPS * 180 + 30) / 3600, 1)
out = f"stargazers-cabin-{hours:g}h.mp4"
print("Building", out, "(takes a few minutes)")
subprocess.run(["bash", "build.sh", str(LOOPS), out], check=True)

dest = f"/content/drive/MyDrive/{DRIVE_FOLDER}"
os.makedirs(dest, exist_ok=True)
print("Copying to Google Drive...")
shutil.copy(out, dest)
drive.flush_and_unmount()
print(f"Done: My Drive / {DRIVE_FOLDER} / {out}")
```
